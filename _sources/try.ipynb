```python
import pandas as pd
import tsfel

# 1. Buat data time series (20 titik)
data = {
    "time": list(range(1, 21)),
    "value": [10,12,11,15,14,13,16,18,17,19,
              20,18,17,16,15,14,13,12,11,10]
}

df = pd.DataFrame(data)

# 2. Simpan ke CSV
df.to_csv("timeseries.csv", index=False)

# 3. Ambil konfigurasi fitur statistik dari TSFEL
cfg = tsfel.get_features_by_domain("statistical")

# 4. Ekstraksi fitur statistik
X = tsfel.time_series_features_extractor(cfg, df["value"])

# 5. Tampilkan hasil
print(X.T)   # .T biar lebih mudah dibaca (transpose)
```
